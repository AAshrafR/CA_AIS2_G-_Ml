```python
'''Task 1:
    Email Domain checking'''

email = "Amit_ml@gmail.std.edu"

# here we are checking if email contains @ or no 
if email.count("@") != 1:
    print("Invalid email")
else:
    # Split the email into username and domain parts
    userName, domainPart = email.split("@")

    # Check if there is a dot after @
    if "." not in domainPart :
        print("Invalid email")
    else:
        last_dot = domainPart.rfind(".") # Find the last occurrence of dot in the domain part
        domain = domainPart[last_dot:]  # Extract the domain ending


        # Check domain ending
        def check_domain(domain):
            if domain.endswith(".com"):
                return "Commercial Domain"
            elif domain.endswith(".edu"):
                return "Educational Domain"
            else:
                return "Other Domain"
print (f"Email: {email}, Domain Type: {check_domain(domain)}")
```

```python
'''task 2: 
    Encode Message: ###!!@mocleW EPGTQ!!!6789
    1_ reverse 1st word
    2_ 2nd word remains same without any change'''

full_message = "###!!@mocleW EPGTQ!!!6789"
text = ""

#extract the spaces and alphabets only from the full message
for i in full_message:
    if i.isalpha() or i == " ":
        text += i

#remove any starting or ending spaces from the text
text = text.strip()

print(f"Original Text : {text}")

#split the text into two words
word1, word2 = text.split()

#reverse the first word
word1 = word1[::-1]
#combine the two words into a single string
text = f"{word1} {word2}"

print(f"Modified Text : {text}")
```

```python
'''task 3: 
    Encode Message: &&&**$gnirtS PLIO!!@1234
    1_ reverse 1st word
    2_ 2nd word replace shifted vowels in the second word: "PLIO": Replace I->E and O->U'''

full_text="&&&**$gnirtS PLIO!!@1234"

#tihs task will be used for task 3&4
def deal_with_input(full_text):
    #extract the spaces and alphabets only from the full message
    filtered_text = ""
    for i in full_text:
        if i.isalpha() or i == " ":
            filtered_text += i

    #remove any starting or ending spaces from the text
    filtered_text = filtered_text.strip()
    #split the text into two words

    return filtered_text

print(f"Original Text : {deal_with_input(full_text)}")
word1, word2 = deal_with_input(full_text).split()

word1 = word1[::-1]
word2 = word2.replace("I", "E")
word2 = word2.replace("O", "U")

text = f"{word1} {word2}"
print(f"Modified Text : {text}")
```

```python
'''task 4: 
    Encode Message: ##$$$@!yalpstcejorp EPUVT****9887
    1_ reverse 1st word
    2_ 2nd word replace shifted vowels in the second word: "EPUVT": Replace E->A, U->O'''

full_text="##$$$@!yalpstcejorp EPUVT****9887"
print(f"Original Text : {deal_with_input(full_text)}")
word1, word2 = deal_with_input(full_text).split()

word1 = word1[::-1]
word2 = word2.replace("E", "A")
word2 = word2.replace("U", "O")

text = f"{word1} {word2}"
print(f"Modified Text : {text}")
```
