```python
from abc import ABC,abstractmethod

class Animal(ABC):

    @abstractmethod
    def move(self):
        pass

class Bird(Animal):
    def move(self):
        print('Birds can fly')

class Cat(Animal):
    def move(self):
        print('Cat can walk')
```

```python
a=Bird()
a.move()
```

```python
class Student:
    #duunded__ constructoe
    def __init__(self,name,age):
        self._name = name #protected
        self.__age = age  #private

    def std_info(self):
        return f'my name is {self._name} and my age is {self.__age}'

    @property
    def name(self):
        return self._name
    
    @property
    def age(self):
        return self.__age
    
    @age.setter
    def age(self,value):
        if value<0:
            raise ValueError('age must be positive')
        self.__age=value



s1= Student('Abdelmasih',24)
print(s1.std_info())
s1.age
```

```python
class Shape:
    def calculate_area(self):
        pass

class Cirle(Shape):
    def __init__(self,radus):
        self.radus=radus

    def calculate_area(self):
        return 3.14*self.radus**2

class Sqare(Shape):
    def __init__(self,radus):
            self.radus=radus
    
    def calculate_area(self):
        return self.radus**2
    
if __name__=='__main__':
    x=Cirle(2)
    y=Sqare(6)
    print(f'Circle Area = {x.calculate_area()}')
    print(f'Sqare Area = {y.calculate_area()}')
    
```
