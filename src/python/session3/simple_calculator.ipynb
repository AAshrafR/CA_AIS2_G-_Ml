```python
def addition(number1, number2):
    return number1 + number2


def subtraction(number1, number2):
    return number1 - number2


def multiplication(number1, number2):
    return number1 * number2


def division(number1, number2):
    return number1 / number2
```

```python
operation = ""

while operation != "0":
    operation = input(
        "Select An Operation:\n"
        "1. Addition (+)\n"
        "2. Subtraction (-)\n"
        "3. Multiplication (*)\n"
        "4. Division (/)\n"
        "0. Exit\n"
        "Enter your choice: "
    )

    if operation == "0":
        print("Goodbye!")
        break

    if operation not in ["1", "2", "3", "4"]:
        print("Invalid operation selected.")
        continue

    number1 = float(input("Enter 1st number: "))
    number2 = float(input("Enter 2nd number: "))

    if operation == "1":
        print(f"Result: {addition(number1, number2)}")

    elif operation == "2":
        print(f"Result: {subtraction(number1, number2)}")

    elif operation == "3":
        print(f"Result: {multiplication(number1, number2)}")

    elif operation == "4":
        if number2 != 0:
            print(f"Result: {division(number1, number2)}")
        else:
            print("Error: Division by zero is not allowed.")
```
