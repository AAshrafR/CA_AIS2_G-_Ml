```python
def train_model(model_name,*datasets,**hyperparameters):
    print(f"Training model: {model_name}")
    print("\nDatasets:")
    for dataset in datasets:
        print('-',dataset)
    print("\nHyperparameters:")
    for key,value in hyperparameters.items():
        print(f"{key}: {value}")

train_model('Fraud detection','train.csv','validation.csv','test.csv',learning_rate=0.01,epochs=100,batch_size=32,optimizer='adam',early_stopping=True,dropout_rate=0.5)
```

```python
transformer={
    'lower': lambda text:text.lower(),
    'upper': lambda text:text.upper(),
    'strip': lambda text:text.strip(),
    'title': lambda text:text.title(),
    'reverse': lambda text:text[::-1],
    'word_count': lambda text:len(text.split()),
    'char_count': lambda text:len(text),
    'remove_spaces': lambda text:text.replace(' ','')
}
```

```python
text= ' machine learning with python '
print(transformer['upper'](text))
```
