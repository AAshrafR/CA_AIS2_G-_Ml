```python
chatbot = {
    "hello": "Hi! How are you?",
    "hi": "Hello! Nice to meet you.",
    "good morning": "Good Morning!",
    "good evening": "Good Evening!",
    "how are you": "I'm doing great, thanks for asking!",
    "what is your name": "My name is ChatBot.",
    "who are you": "I'm a simple Python chatbot.",
    "bye": "Goodbye! Have a nice day.",
    "thanks": "You're welcome!",
    "thank you": "My pleasure!",
    "good": "Glad to hear that!",
    "sad": "I'm sorry to hear that. I hope things get better.",
    "happy": "That's great! Keep smiling!",
    "help": "I'm here to help you.",
    "default": "Sorry, I don't understand."
}
```

```python
user = input("You: ").lower()

if user in chatbot:
    print("Bot: ", chatbot[user])
else:
    print("Bot: ", chatbot["default"])
```

```python
number1 = int(input("Enter a number: "))
number2 = int(input("Enter another number: "))
opration = input("Enter an operation (1 for +,2 for -,3 for *, 4 for /): ")
if opration == "1":
    result = number1 + number2
    print(f"Result: {result}")
elif opration == "2":
    result = number1 - number2
    print(f"Result: {result}")
elif opration == "3":
    result = number1 * number2
    print(f"Result: {result}")
elif opration == "4":
    if number2 != 0:
        result = number1 / number2
        print(f"Result: {result}")
    else:
        print("Error: Division by zero is not allowed.")
else:
    print("Invalid operation selected.")
```
