# Task 1 number multiplication table

```python
def get_multiplication_table(number:int) -> None:
    '''
    Print the multiplication table of a given number from 1 to 10.

    Args:
        number (int): The number whose multiplication table will be printed.

    Returns:
        None: This function only prints the multiplication table.

    Example:
        multiplication_table(5)

        you will get a multiplication table from 1 to 10:
            5 x 1 = 5
            ......
            5 x 10 = 50
    '''
    for i in range(1, 11):
        print(f'{number} x {i} = {number * i}')


get_multiplication_table(5)
```

# Task 2 twin prime pairs less than 1000

```python
def is_prime(number: int) -> bool:
    '''
    Check whether a given number is prime.

    Args:
        number (int): The number to check.

    Returns:
        bool: True if the number is prime otherwise False.

    Example:
        is_prime(7)
        True
    '''
    if number < 2:
        return False

    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False

    return True
```

```python
def get_twin_prime(limit: int) -> None:
    '''
    Print all twin prime pairs less than the given limit.

    Args:
        limit (int): The upper limit for finding twin primes.

    Returns:
        None: This function only prints the twin prime pairs.

    Example:
        print_twin_primes(8)

        Output:
            (3 AND 5)
            (5 AND 7)
    '''
    for number in range(3, limit, 2):
        if is_prime(number) and is_prime(number + 2): # & يعني لو الاتنين تروووووووووووووووووووووووووو
            print(f"({number} AND {number + 2})")


get_twin_prime(1000)
```

# Task 3 number factors

```python
def prime_factors(number: int) -> list[int]:
    '''
    Find the prime factors of a given number.

    Args:
        number (int): The number whose prime factors will be found.

    Returns:
        list[int]: A list containing the prime factors of the number.

    Example:
        prime_factors(56)
        [2, 2, 2, 7]
    '''
    factors = []
    divisor = 2

    while number > 1:
        while number % divisor == 0:
            factors.append(divisor)
            number //= divisor

        divisor += 1

    return factors


prime_factors(56)
```

# Task 4 dicemal to binary

```python
def dec_to_binary(number: int) -> str:

    '''
    Convert a decimal number to its binary representation.

    Args:
        number (int): The decimal number to convert.

    Returns:
        str: The binary representation of the number.

    Example:
        dec_to_binary(9)
        '1001'
    '''

    result = ''

    while number > 0:
        if number % 2 == 0:
            result = '0'+ result
        else:
            result = '1' + result

        number //= 2

    return result


dec_to_binary(13)
```

# Task 5 perfect number

```python
def is_perfect(number: int) -> bool:
    '''
    Check whether a given number is a perfect number.

    Args:
        number (int): The number to check.

    Returns:
        bool: True if the number is perfect, otherwise False.

    Example:
        is_perfect(28)
        True

        is_perfect(10)
        False
    '''
    divisor_sum = 0

    for divisor in range(1, number):
        if number % divisor == 0:
            divisor_sum += divisor

    return divisor_sum == number
```

```python
def print_perfect_numbers(start: int, end: int) -> list[int]:
    '''
    Print all perfect numbers within a given range.

    Args:
        start (int): The starting value of the range.
        end (int): The ending value of the range.

    Returns:
        list[int]: This function returns a list of perfct numbers in the givin range.

    Example:
        print_perfect_numbers(1, 100)

        Output:
            0
            6
            28
    '''
    perfect_lst=[]
    for number in range(start, end + 1):
        if is_perfect(number):
            perfect_lst.append(number)
    return perfect_lst


print_perfect_numbers(0,100)
```

```python
def get_lst_items_sum(lst: list[int]) -> int:
    '''
    Convert a reversed list of digits into an integer.

    Args:
        lst (list[int]): A list of digits stored in reverse order.

    Returns:
        int: The integer represented by the digits.

    Example:
        get_lst_items_sum([2, 4, 3])
        342
    '''
    total = 0

    for i in range(len(lst)):
        total += lst[i] * (10 ** i)

    return total


def get_addition_result(lst1: list[int], lst2: list[int]) -> list[int]:
    '''
    Add two numbers represented by reversed lists of digits.

    Args:
        lst1 (list[int]): The first number represented in reverse order.
        lst2 (list[int]): The second number represented in reverse order.

    Returns:
        list[int]: The sum represented as a reversed list of digits.

    Example:
        get_addition_result([2, 4, 3], [5, 6, 4])
        [7, 0, 8]
    '''
    total = get_lst_items_sum(lst1) + get_lst_items_sum(lst2)

    result_lst = [int(x) for x in str(total)][::-1]

    return result_lst


lst1 = [2, 4, 3]
lst2 = [5, 6, 4]

get_addition_result(lst1, lst2)
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def get_linked_list_sum(head: ListNode) -> int:
    '''
    Convert a reversed linked list of digits into an integer.

    Args:
        head (ListNode): The head of the linked list.

    Returns:
        int: The integer represented by the linked list.

    Example:
        2 -> 4 -> 3 -> None
        342
    '''
    total = 0
    power = 0
    current = head

    while current:
        total += current.val * (10 ** power)
        power += 1
        current = current.next

    return total


def create_linked_list(digits: list[int]) -> ListNode:
    '''
    Create a linked list from a list of digits.

    Args:
        digits (list[int]): A list of digits in reverse order.

    Returns:
        ListNode: The head of the created linked list.

    Example:
        create_linked_list([7, 0, 8])

        Output:
            7 -> 0 -> 8 -> None
    '''
    head = ListNode(digits[0])
    current = head

    for digit in digits[1:]:
        current.next = ListNode(digit)
        current = current.next

    return head


def add_two_numbers(lst1: list[int], lst2: list[int]) -> ListNode:
    '''
    Add two numbers represented by reversed lists of digits
    and return the result as a linked list.

    Args:
        lst1 (list[int]): The first number represented in reverse order.
        lst2 (list[int]): The second number represented in reverse order.

    Returns:
        ListNode: The head of the result linked list.

    Example:
        add_two_numbers([2, 4, 3], [5, 6, 4])

        Output:
            7 -> 0 -> 8 -> None
    '''
    linkedlst1 = create_linked_list(lst1)
    linkedlst2 = create_linked_list(lst2)

    total = get_linked_list_sum(linkedlst1) + get_linked_list_sum(linkedlst2)

    digits = [int(x) for x in str(total)][::-1]

    return create_linked_list(digits)
```
